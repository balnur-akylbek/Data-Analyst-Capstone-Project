#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
# 1. Проверяем количество пропущенных значений в указанных столбцах
critical_cols = ['Employment', 'JobSat', 'RemoteWork']
print("Пропуски до обработки:")
print(df[critical_cols].isnull().sum())

# 2. Заполняем пропуски самым частым значением (модой) для каждого столбца
for col in critical_cols:
    majority_value = df[col].mode()[0]
    df[col] = df[col].fillna(majority_value)

# 3. Проверяем результат, чтобы убедиться, что пропусков не осталось (выведет 0)
print("\nПропуски после обработки:")
print(df[critical_cols].isnull().sum())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
# 1. Очищаем YearsCodePro и переводим в числа
df['YearsCodePro_Num'] = pd.to_numeric(df['YearsCodePro'].replace({'Less than 1 year': '0', 'More than 50 years': '51'}), errors='coerce').fillna(0)

# 2. Создаем диапазоны опыта (0-5, 5-10, 10-20, >20 лет)
bins = [-1, 5, 10, 20, 100]
labels = ['0-5 years', '5-10 years', '10-20 years', '>20 years']
df['ExperienceRange'] = pd.cut(df['YearsCodePro_Num'], bins=bins, labels=labels)

# 3. Переводим текстовую удовлетворенность JobSat в числа (от 1 до 5)
# Примечание: Проверьте точные названия категорий в вашем df['JobSat'].unique(), если они отличаются
sat_mapping = {
    'Very dissatisfied': 1, 'Slightly dissatisfied': 2,
    'Neither satisfied nor dissatisfied': 3,
    'Slightly satisfied': 4, 'Very satisfied': 5
}
df['JobSat_Numeric'] = df['JobSat'].map(sat_mapping).fillna(3) # Реальные пропуски заполняем нейтральным значением

# 4. Считаем медиану JobSat для каждого диапазона опыта
median_sat = df.groupby('ExperienceRange')['JobSat_Numeric'].median().reset_index()
print(median_sat)

# 5. Визуализация отношения с помощью Bar Plot
plt.figure(figsize=(8, 5))
sns.barplot(data=median_sat, x='ExperienceRange', y='JobSat_Numeric', palette='Blues_d')
plt.title('Median Job Satisfaction by Experience Range')
plt.ylabel('Median JobSat Score (1-5)')
plt.ylim(1, 5)
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
plt.figure(figsize=(10, 5))
# Строим count plot распределения ответов удовлетворенности
sns.countplot(data=df, y='JobSat', order=df['JobSat'].value_counts().index, palette='viridis')
plt.title('Distribution of Job Satisfaction Levels')
plt.xlabel('Count')
plt.ylabel('Job Satisfaction')
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
# 1. График распределения удаленной работы (Count plot)
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='RemoteWork', palette='Set2')
plt.title('Distribution of Remote Work Preferences')
plt.xticks(rotation=45)
plt.show()

# 2. Кросс-табуляция (Cross-tabulate) между удаленной работой и типом занятости
# Примечание: Если вам нужно сопоставить с конкретными ролями, замените 'Employment' на нужный столбец (например, 'DevType')
ct = pd.crosstab(df['Employment'], df['RemoteWork'], margins=True)
print("Кросс-таблица зависимостей:")
print(ct)
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
# 1. Очищаем данные от пропусков и разделяем языки программирования
df_langs = df[['Country', 'LanguageHaveWorkedWith']].dropna()
df_langs['Language'] = df_langs['LanguageHaveWorkedWith'].str.split(';')
df_langs = df_langs.explode('Language')

# 2. Выбираем топ-3 страны для анализа (например: United States, United Kingdom, Germany)
# Вы можете заменить названия стран на любые другие из вашего df['Country'].unique()
top_countries = df['Country'].value_counts().head(3).index
df_filtered = df_langs[df_langs['Country'].isin(top_countries)]

# 3. Считаем топ-5 популярных языков для этих регионов
top_langs = df_filtered['Language'].value_counts().head(5).index
df_filtered = df_filtered[df_filtered['Language'].isin(top_langs)]

# 4. Визуализация в виде сгруппированного Bar Plot
plt.figure(figsize=(12, 6))
sns.countplot(data=df_filtered, x='Language', hue='Country', order=top_langs, palette='muted')
plt.title('Top Programming Languages by Region')
plt.xlabel('Programming Language')
plt.ylabel('Count')
plt.legend(title='Region/Country')
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
# 1. Приводим стаж к числовому типу, обрабатывая строки 'Less than 1 year' и т.д.
df['YearsCodePro_Numeric'] = pd.to_numeric(
    df['YearsCodePro'].replace({'Less than 1 year': '0', 'More than 50 years': '51'}), 
    errors='coerce'
).fillna(0)

# 2. Приводим оценки к числовому типу и убираем пропуски для графика
df_corr = df[['YearsCodePro_Numeric', 'JobSatPoints_1']].dropna()
df_corr['JobSatPoints_1'] = pd.to_numeric(df_corr['JobSatPoints_1'], errors='coerce')

# 3. Визуализация связи с помощью Scatter Plot с линией тренда
plt.figure(figsize=(10, 6))
sns.regplot(data=df_corr, x='YearsCodePro_Numeric', y='JobSatPoints_1', 
            scatter_kws={'alpha':0.3, 'color':'blue'}, line_kws={'color':'red'})
plt.title('Correlation Between Experience and Job Satisfaction Points')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Job Satisfaction Points')
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
# 1. Создаем кросс-таблицу процентного соотношения (нормализация по строкам)
edu_employment_pct = pd.crosstab(df['EdLevel'], df['Employment'], normalize='index') * 100

# 2. Строим тепловую карту (Heatmap)
plt.figure(figsize=(12, 8))
sns.heatmap(edu_employment_pct, annot=True, fmt=".1f", cmap="YlGnBu", cbar_kws={'label': 'Percentage (%)'})
plt.title('Correlation Between Educational Background and Employment Type (%)')
plt.xlabel('Employment Type')
plt.ylabel('Education Level (EdLevel)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
